# Wavefronts, optics, and propagators

This tutorial follows a `Wavefront` as optical layers modify its complex field and propagators move it between sampled planes. It explains the relationship between the phasor, wavelength, physical `GridSpec`, amplitude, phase, intensity, and power, then shows how these values and metadata change through an optical system.

The tutorial has five stages:

1. **Construct a wavefront.** Create monochromatic and chromatic wavefronts on a physical pupil grid, inspect their amplitude, phase, intensity, power, wavelength axes, and coordinates, and apply direct OPD, phase, tilt, and normalisation operations. Leading wavelength axes are handled by dLux rather than manually vectorised by the user.
2. **Apply an optic.** Combine transmission, physical OPD, and phase in one `Optic`, then compare the wavefront before and after application. Transmission multiplies the field amplitude, OPD contributes the chromatic phase $2\pi\,\mathrm{OPD}/\lambda$, and phase is already expressed in radians.
3. **Propagate to and from focus.** Compare Fraunhofer MFT propagation on an explicit focal `GridSpec` with FFT propagation controlled by a `ResizeSpec`. Match their realised sampling before comparing intensities and residuals, then demonstrate reverse focal propagation with `inverse=True`, including the effects of padding and cropping on round-trip accuracy.
4. **Model other propagation regimes.** Use `Fresnel` for defocused focal planes, `FreeSpace` for angular-spectrum propagation between physical planes, and a small `ABCDPropagator` lens system for a composed paraxial model. Explain the relevant output grids, physical units, supported numerical methods, and why an ABCD system has no generic inverse flag.
5. **Inspect a complete optical system.** Combine the optic and propagators in an `OpticalSystem` and use `debug_propagate_mono()` to inspect the wavefront after each layer. The final comparison should make the central contract explicit: every optical layer receives and returns a wavefront; an optic changes its field at one plane, while a propagator changes both the field and its spatial grid.

The closing guide maps physical problems to `Fraunhofer`, `Fresnel`, `FreeSpace`, and `ABCDPropagator`, and distinguishes explicit `GridSpec` output sampling from FFT padding and cropping through `ResizeSpec`. The tutorial contains no optimisation or inference; its validation comes from physical units, realised grids, intermediate fields, matched-method agreement, and propagation residuals.
