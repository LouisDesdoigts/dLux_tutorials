# High-dimensional detector calibration

This tutorial builds a dithered calibration observation and jointly recovers a high-dimensional detector response with a small set of optical and observational nuisance parameters. A stable optical source is observed through spatial pixel response, jitter, background, and other detector effects already supported by dLux.

The workflow introduces a custom observation class, shared and exposure-dependent parameter paths, regularised pixel-response inference, staged optimisation, and held-out validation. Diagnostics compare the injected and recovered response map, parameter histories, and model/data/z-score images before and after calibration. Nonlinear gain is intentionally excluded until dLux provides a reviewed detector-layer contract for it.
