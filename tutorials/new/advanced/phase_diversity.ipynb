# JWST phase-diversity phase retrieval

This guide uses a JWST-like segmented aperture and an implicit Fourier OPD basis to recover a wavefront from focused and deliberately defocused observations. The same pupil and source parameters are shared across both images, while the known diversity changes the information available to the fit.

The workflow is deliberately compact: build and inspect the JWST model, inject a spatial wavefront, simulate focused and defocused noisy images, recover the shared OPD with staged SGD, and compare the initial and recovered image and pupil residuals. The tutorial explains the focused-image ambiguity, the role and scale of the diversity, and the practical sampling and optimisation choices. It does not calculate Hessians, Fisher matrices, or covariance estimates; those are unnecessarily memory-intensive for a pixel-scale implicit Fourier basis.
