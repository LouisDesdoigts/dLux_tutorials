# Modelling the MagAO-X coronagraph

This tutorial builds a structurally faithful MagAO-X classical Lyot coronagraph from the Magellan entrance pupil through the science focal plane. The model uses the documented entrance or bump pupil mask, physical opaque focal-plane mask, measured Lyot stop, and the real sampling and dimensions supplied by MagAO-X and HCIPy. Instrument files and published dimensions are treated as sourced model inputs rather than replaced by generic circular masks.

The optical path is Magellan pupil → entrance mask → focal propagation → focal-plane mask → inverse propagation → Lyot stop → science focus. We inspect the amplitude and phase at every accessible plane, establish agreement between explicit propagation and the compact `SoummerFPM` algorithm, then evaluate polychromatic on-axis suppression, off-axis throughput, and sensitivity to pupil-, focal-mask-, and Lyot-stop alignment. A non-Optax optimiser is used for a physically motivated alignment or performance problem.

The MagAO-X coronagraph handbook and HCIPy Magellan aperture implementation define the starting fidelity and validation targets. A later extension may add the instrument's vAPP, PIAACMC, polarimetric behaviour, or self-coherent-camera path after the classical Lyot model is established and the required dLux layers have reviewed contracts.

Sources: [MagAO-X coronagraph handbook](https://magao-x.org/docs/handbook/observers/coronagraphs.html), [HCIPy Magellan aperture](https://docs.hcipy.org/dev/api/hcipy.aperture.make_magellan_aperture.html), and [HCIPy broadband Magellan example](https://docs.hcipy.org/0.3.0/tutorials/BroadbandTelescopePSF/BroadbandTelescopePSF.html).
