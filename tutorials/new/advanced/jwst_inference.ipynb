# JWST NIRCam stellar-field inference

This flagship tutorial constructs a high-fidelity JWST NIRCam imaging model, simulates a vectorised stellar field, and jointly recovers the position and flux of every star. Instrument fidelity includes a realistic JWST pupil and OPD, an actual NIRCam filter bandpass and detector sampling, and supported wavelength-, field-, and detector-dependent effects validated where possible against STPSF, WebbPSF, and the existing dLuxWebbpsf implementation. The tutorial states explicitly which effects are represented and which are omitted.

The field contains stars with different positions, log fluxes, and fixed effective temperatures, producing source-dependent blackbody spectra across the observing band. Multiple noisy exposures may include dithers or detector rotation. Initial catalogue positions come from correlation or matched filtering and initial fluxes from local photometry. A custom observation class then supports joint astrometric and photometric optimisation, with optional recovery of a small identifiable instrumental nuisance set such as jitter or low-order wavefront drift.

The final diagnostics compare the injected and recovered catalogues, astrometric and photometric errors, completeness and source confusion, parameter histories, and model/data/z-score images. Effective-temperature inference and more complete detector calibration remain later extensions rather than being folded into the first complete fit.
