# Apertures and aperture builders

This tutorial explains how dLux turns physical geometry and OPD definitions into sampled arrays, dense optical layers, compact pasted pupils, and genuinely sparse optics. It focuses on the construction framework rather than cataloguing named telescope templates, which are covered separately in the prebuilt tutorial. By the end, the reader should be able to construct a custom aperture, attach a supported OPD basis, transform its coordinates, and select an appropriate dense or sparse representation.

The central construction path is `GridSpec + Shape + OPDDef → ApertureBuilder → arrays from .build() or an Optic from __call__()`. The tutorial has five stages:

1. **Sample an explicit aperture.** Combine a `Circle` primary with central and `Spider` obscurations, evaluate it on a physical `GridSpec`, and compare the primary support with the final transmission. Explain grid promotion, physical coordinates, hard-edge oversampling, downsampling, and why support is not simply every non-zero transmission pixel.
2. **Add a supported OPD basis.** Attach a `ZernikeDef`, compare mutually exclusive `nolls` and `orders`, and use `Norm` to normalise each basis vector directly over the aperture support. Contrast `.build()`, which returns explicit transmission and basis arrays, with calling the builder, which creates an `Optic` and initialises its `Basis` from `coeffs`, a random key, or zeros.
3. **Transform and soften geometry.** Introduce representative `Circle`, `Rectangle`, `RegularPolygon`, `Spider`, `Complement`, and `TransformedShape` objects. Compare the default hard boundary with `Soft`, distinguish soft differentiable edges from construction-time oversampling, and apply a `CoordTransform` to rotate or translate a builder without manually manipulating sampled arrays.
4. **Build dense, pasted, and sparse pupils.** Construct a generic repeated-subaperture model with `SparseApertureBuilder`, explaining local versus global obscurations, explicit `(x, y)` centres, and per-aperture OPD coefficients. Compare the default dense optic with `sparse=True`, then explain how ideal `SegmentedHex` pupils and `PastedBasis` evaluate compact stamps. Demonstrate numerical agreement while documenting the `scan` memory-efficient and `scatter` parallel construction paths and the dense fallback required by general transformations.
5. **Define a reusable aperture.** Wrap the initial three-spider geometry in a small `ApertureBuilder` subclass. Show that custom builders normally assemble existing shapes and OPD definitions while inheriting the standard sampling, support, transformation, JIT, and optical-layer contracts; they should not reimplement coordinate generation or downsampling. Use the resulting builder in a minimal optical system and generate a PSF.

The final summary maps common tasks to their APIs: `.build(grid)` for explicit arrays, `return_support=True` for support, `builder(grid)` for an optical layer, `ZernikeDef` for supported modes, `coeffs` or `key` for initialisation, `transform` for coordinate distortion, `SparseApertureBuilder` for repeated local pupils, `SegmentedHex` and `PastedBasis` for compact ideal segments, and `sparse=True` for genuinely sparse propagation. The tutorial contains no optimisation or inference; its purpose is to make the aperture contracts predictable before the prebuilt tutorial applies them to instrument templates.
